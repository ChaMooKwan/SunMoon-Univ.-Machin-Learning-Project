# 정답레이블: Aspect, SentimentPolarity

```python
# 'Training_전체통합' 와 'Validation_전체통합' 파일을 통합
import pandas as pd

df1 = pd.read_csv('Training_전체통합.csv')
df2 = pd.read_csv('Validation_전체통합.csv')
df = pd.concat([df1, df2])
```

```python
df = pd.concat([df, pd.get_dummies(df['Aspect'], dtype = int)], axis=1)
```

```python
df.info()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 예시 데이터프레임 생성 (A 컬럼과 B 컬럼이 레이블인 경우)
# df['label_A'], df['label_B'] 가 있다고 가정합니다.

# 1. 두 컬럼을 합친 임시 stratify 기준 생성
# 예: A=1, B=0 -> "1_0" 이라는 하나의 범주가 됨
stratify_key = df['Aspect'].astype(str) + "_" + df['SentimentPolarity'].astype(str)

# 2. train_test_split 실행
train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=stratify_key  # 합쳐진 키를 기준으로 비율을 맞춤
)

# 확인: 원래 데이터와 분리된 데이터의 조합 비율이 일치하는지 체크
print(train_df.groupby(['Aspect', 'SentimentPolarity']).size() / len(train_df))
print(test_df.groupby(['Aspect', 'SentimentPolarity']).size() / len(test_df))
```

```python
train_df.to_csv('Training_비율통합.csv', index=False, encoding='utf-8-sig')
test_df.to_csv('Validation_비율통합.csv', index=False, encoding='utf-8-sig')
```
